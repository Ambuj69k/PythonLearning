```python
data=('col1,col2,col3\n'
'x,y,1\n'
'a,b,2\n'
'c,j,4')

"""simply agar hum data define karte and when we check type of the data wo simply string and agar direct 
hum read karne ka try kare to wo error throw karega is case me hume particular data jo hai wo memory me loaded chaiye
hota hai aur us case me IO use karte hai"""
type(data)
```

```python
from io import StringIO
```

```python
dataforporcess=StringIO(data)
```

```python
pss=pd.read_csv(dataforporcess)
```

```python
pss
```

```python
"""
agar simply hum specific columns use karna hai to hum simply 
usecols use karenge """

prr=pd.read_csv(dataforporcess,usecols=['col1','col2'])
```

```python
prr
```

```python
prr.to_csv('new_csv') #using this we can create a new csv file and save it  in the current directory
```

```python
"""while reading the csv file we can also specify the datatype should be considered for each column
using dtype
"""

df=pd.read_csv(StringIO(data),dtype='object')
df.info
```

```python
"""agar hume aisa karna hai to hum simply isko bhi aisa kar sakte hai ki hum specific columns ke liye datatype bhi set kar sakte hai ye karne ke liye hume simply dictionary use kar sakte hai specific column ke liye specific value set karne ke liye"""

df=pd.read_csv(StringIO(data),dtype={'col1':'str','col2':'str','col3':'int'})
```
