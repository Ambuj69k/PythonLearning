```python
##lets define a functoin for welcoming new user

def welcome_user():
    """
    This function will welcome the user
    
    this function will return a message for welcoming the user
    """
    return "Welcome to our application!"
```

```python
welcome_user()
```

```python
"""postional argument and keyword argument
positional argument is the argument which is passed to the function in the order in which it is  """

def user(*args,**kwargs):
    """
    This function will take positional and keyword arguments
    
    this function will return a message for welcoming the user
    """
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)


user("John", "Doe", age=30, city="New York")
```

```python
lst=["jhon","doe"]
dic={"age":30,"city":"New York"}

user(lst,dic)
```

```python
user(*lst,**dic)
```
