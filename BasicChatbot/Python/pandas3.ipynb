````python
import pandas as pd
from io import StringIO
````

````python
"""This is a simple example of pandas using json """

data='{"employee_name":"james","email":"james123@gmail.com","job_profile":{"title1":"Team Lead","title2":"sr.developer"}}}'
````

````python
type(data)
````

````python
pd.read_json(StringIO(data),orient='records') #using this we can read the json data and convert it into a pandas dataframe
````

````python
pd.json_normalize(data) #using this we can read the json data and convert it into a pandas dataframe
````

````python
"""to use split as orient we need to use json data into a specific format like this
{
  "columns": ["col_1", "col_2"],
  "index": ["row_1", "row_2"],
  "data": [
    ["val_1", "val_2"],
    ["val_3", "val_4"]
  ]
}
```

### Key Components
* **`columns`**: A list containing the column labels of the DataFrame.
* **`index`**: A list containing the row labels or index values of the DataFrame.
* **`data`**: A 2D array (list of lists) where each inner list represents a single row of values corresponding to the order of the columns.

<FollowUp>
If you want, tell me:
* Are you **reading** or **writing** the JSON data?
* Do you have a **sample of your data**?

I can show you the exact code snippet for your use case.
</FollowUp>
"""
````
