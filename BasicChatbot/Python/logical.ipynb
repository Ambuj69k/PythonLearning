```python
##Program to find Prime Number 
num= int(input("enter your number"))
if num<2:
    print("Not a prime number")
else:
    is_prime=True
    for i in range(2,num):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        print("Number is prime Number")
    else:
        print("Number is not prime Number")
```
