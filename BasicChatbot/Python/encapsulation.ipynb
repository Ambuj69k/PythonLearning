```python
"""
kisi bhi variable ko agar private banana hota hai to jab wo constructor me set ye intialise hota hai to 
simply hume kya karna hai self.__name=name
and protected karne ke liye instead of double underscore we use single under score simply

"""

class Person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
    def display_inf0(self):
        print(f"the name is {self.__name} and age is {self.__age}")

person=Person("ambuj",49)
person.display_inf0()

```
