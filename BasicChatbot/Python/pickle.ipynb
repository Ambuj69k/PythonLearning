```python
"""the pickle module implements binary protocols for serializing and de serializing a pyton object struucure"""
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset("iris")
```

```python
df
```

```python
import pickle
```

```python
filename='file.pkl'#pickle file ka extension hamesha .pkl hota hai

pickle.dump(df,open(filename,'wb'))#wb ka matlab h write binary mode me file ko open karna serialize karne ke liye
```

```python
#unserialize karne ke liye hum simply pickle.load ka use karenge
df=pickle.load(open(filename,'rb'))#rb ka matlab h read binary mode me file ko open karna unserialize karne ke liye
```
