```python
import pandas as pd
import requests
from io import StringIO
```

```python
"""html contnent ko read karne ke liye pandas simply kya karta hai wo html content me se tr and th ko read karta
hai to jo bhi hum data de html ka wo usme tr and th check karta hai and read karke usko return karta hai
"""

#wikipedia blocks python's default downloader (403), so fetch the page with a browser user-agent first
url="https://en.wikipedia.org/wiki/Jeff_Bezos"
response=requests.get(url,headers={"User-Agent":"Mozilla/5.0"})

html=pd.read_html(StringIO(response.text))
len(html)
```

```python
#agar specific table hume uthana hai to match simply jo hai uska use karke hum kar sakte hai simply hume bas
html=pd.read_html(StringIO(response.text),match="Billions")
```
