intersection_update jo hota hai wo simply kya karta hai intersection ka jo value hai suppose s.intersection(b) kiya to iska value update nhi hota bas wo calculate karke dikhata hai but agar hum intersection update use kare to simply kya hoga simply wo update hota hai

```python
set_var=set((1,2,3,4,5,5))
set_var2=set((2,5,6,3))

set_var2.intersection_update(set_var)
set_var2.add(8)
set_var2.difference(set_var)
```

### Dictionary
simply key value pair use karta hai simply agar value agar nikaalni hoo to simply key ka use karte hai 
dictionary me value hum change kar sakte hai to dictionary jo hai wo mutable hai 

```python
#dictionary

my_dict={"car": "Audi","car2":"bmw","car3":"Mercidies benz"}
my_dict["car"]
```

```python
for i in my_dict:
    print(i)
"""by default jo bhi hum iterate karte hai wo keys ke upar simply iterate karta hai
by default aur ek example hai agar hume values ke upar iterate karna hai for loop me hi hum mentions kar sakte hai """
```

```python
for i in my_dict.items():
    print(i)
```
