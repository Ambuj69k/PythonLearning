```python
# return the length of the string using lambda function
length= lambda s:len(s)
length("hello")
```

```python
#convert the list of integers to their corresponding square values
lis=[1,2,3,4,5,6,7,8,9]
squared= lambda lis: [i*i for i in lis]
squared(lis)
```

```python
#filter out even numbers from a list
lis=[1,2,3,4,5,6,76,7,7,43]
even_li= lambda lis: [i for i in lis if i%2==0]
even_li(lis)
```

```python
# sort a list of strings based on their alpabetical char and lenght

fruits=["apple",'banana','cherry','date','elderberry']

sorted(fruits,key=lambda x:len(x))
```

```python
#sorting list of dictionaries based on the specific key
employees = [
    {
        "name": "Rahul",
        "age": 25,
        "department": "IT"
    },
    {
        "name": "Priya",
        "age": 28,
        "department": "HR"
    },
    {
        "name": "Amit",
        "age": 30,
        "department": "Finance"
    },
    {
        "name": "Sneha",
        "age": 26,
        "department": "IT"
    }
]
```

```python
sorted(employees,key=lambda x:(x['age'],x['name']))
```
