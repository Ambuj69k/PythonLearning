```python
lst=[1,2,3,4,5,6]

even_sum=0
odd_sum=0

for i in lst:
    if i%2==0:
        even_sum=even_sum+i
    else:
        odd_sum=odd_sum+i

print(f"Sum of Even no is: {even_sum}")
print(f"sum of odd no is : {odd_sum}")
```
